```python
#H1 Write a function to calculate the sum of numbers from 1 to N
def sum_of_numbers(n):
  sum = 0
  for i in range(1, n+1):
    sum += i
  return sum

n = int(input("Enter a number : "))
print(f"Sum of 1 to {n} is : {sum_of_numbers(n)}")
```

```python
#H2 Write a function that returns the square of a number.

def sq(x):
  return x**2

n = int(input())
print(sq(n))
```

```python
#H3 Write a function that checks whether a number is even/odd.

def is_even(y):
  if (y % 2 == 0):
    return True
  return False

y = int(input())
if(is_even(y)):
  print(f"{y} is even")
else:
  print(f"{y} is odd")
```

```python
#H4 Write a function that returns the larger number from two given numbers.
def largest_of_two(x, y):
  if(x > y):
    return x
  return y

x = int(input())
y = int(input())
print(f"largest between {x} and {y} is : {largest_of_two(x, y)}")
```

```python
#H5 Write a Python program to create a list of 10 integers. Then perform the following operations:
  #Append an element to the list.
  #Insert an element at a specific position.
  #Remove an element from the list.
  #Pop an element from a specific position.


list_1 = [20, 10, 30, 25, 40, 15, 80, 90, 50, 100]

print(list_1)

list_1.append(50)
print(list_1)

list_1.insert(2, 100)
print(list_1)

list_1.remove(30)
print(list_1)

popped_item = list_1.pop(3)  # pop 25
print(popped_item)
print(list_1)
```

```python
#H6 Write a program to find all unique elements in the list.

list_2 = [70, 10, 50, 100, 20, 60, 30, 80, 40, 100, 90, 50, 10]

print(list_2)

list_2 = list(set(list_2))

print(list_2)
```

```python
#H7 Write a Python program to calculate the mean of a list of numbers.

list_2 = [70, 10, 20, 60, 30, 80, 40, 100, 90, 50]

sum = 0

length = len(list_2)

for i in range(0, length):
  sum += list_2[i]

avg = sum / length
print(avg)
```

```python
#H8 Classify Numbers into Even and Odd Lists

list_3 = [1, 2, 3, 4, 5, 6, 7, 8, 9]
even_list = []
odd_list = []

for x in list_3:
  if x%2 == 0:
    even_list.append(x)
  else:
    odd_list.append(x)

print(even_list)
print(odd_list)
```

```python
# H9: Write a program to find the key in a dictionary that has the maximum value.
# Input: {‘a’: 10, ‘b’: 20, ‘c’: 15}
# Output: ‘b’

nums = {'a' : 10, 'b' : 20, 'c' : 15}
mx = max(nums, key = nums.get)
print(mx)
```

```python
# H10: Write a program to create a dictionary where the keys are
# numbers from 1 to n and the values are their squares.
# Input: n = 5
# Output:
# { 1: 1, 2: 4, 3: 9, 4: 16, 5: 25 }

sq = {}

n = int(input())

for i in range(1, n+1):
  sq[i] = i**2

print(sq)
```

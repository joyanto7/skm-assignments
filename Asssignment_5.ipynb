```python
# target variable --> airconditiong
print("Target variable Distribution:")
print(df['airconditioning'].value_counts())
```

```python
# 2. Separate X (features) and y (target = 'airconditioning')

X = df.drop('airconditioning', axis=1)
y = df['airconditioning']

```

```python
# 3. Label Encode: mainroad, guestroom, basement, hotwaterheating, prefarea

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df['mainroad'] = label_encoder.fit_transform(df['mainroad'])
df['guestroom'] = label_encoder.fit_transform(df['guestroom'])
df['basement'] = label_encoder.fit_transform(df['basement'])
df['hotwaterheating'] = label_encoder.fit_transform(df['hotwaterheating'])
df['prefarea'] = label_encoder.fit_transform(df['prefarea'])
```

```python
# 4. One-Hot Encode: furnishingstatus

furnish_encoded = pd.get_dummies(df['furnishingstatus'], prefix = 'furnishing', dtype = int)
df = pd.concat([df, furnish_encoded], axis = 1)
df.drop('furnishingstatus', axis = 1, inplace = True)
df.head(10)
```

```python
# 5. Train-test split (80% train, 20% test) — remember stratify=y

X = df.drop('airconditioning', axis=1)
y = df['airconditioning']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    stratify=y,       # keep the same Outcome ratio in train and test
    random_state=30   # Same result every time
)
```

```python
# 6. Scale the numeric features with StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 7. Run: train a LogisticRegression model, apply cross validation( 5 fold, 10 fold) and print the accuracy

from sklearn.linear_model import LogisticRegression

#create model
model = LogisticRegression(max_iter=1000)

#Train the model
model.fit(X_train_scaled, y_train)

#check accuracy on training data (optional)
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")

#check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```

## Cross validation

```python
# 5 fold
from sklearn.model_selection import cross_val_score

# Do 5-fold cross-validation
scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=5             # 5 mini-tests
)


print("5 mini-test scores:")
for i in range(5):
    print(f"  Test {i+1}: {scores[i]:.1%}")

print(f"\nAverage: {scores.mean():.1%}")
print(f"This means our model is {scores.mean():.1%} accurate!")
```

```python
# 10 fold
scores10 = cross_val_score(
    model,
    X_train_scaled,
    y_train,
    cv = 10
)

print("10 mini-test scores:")
for i in range(10):
    print(f"  Test {i+1}: {scores10[i]:.1%}")

print(f"\nAverage: {scores10.mean():.1%}")
```
